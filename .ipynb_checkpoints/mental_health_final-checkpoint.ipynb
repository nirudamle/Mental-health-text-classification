# Mental Health Text Classification
### 4-Class: Suicidal | Depression | Anxiety | Normal

```python
import pandas as pd
import numpy as np
import re
import math
import nltk
import pickle
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.utils.class_weight import compute_class_weight

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, Input
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

print('All libraries imported successfully!')
```

---
## 3. Text Preprocessing

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    text   = text.lower()
    text   = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return ' '.join(tokens)

sample = "I have been feeling extremely anxious and depressed lately!!!"
print('Original  :', sample)
print('Processed :', preprocess_text(sample))
```

```python
print('Preprocessing training set...')
train_df['clean_text'] = train_df['text'].apply(preprocess_text)
print('Training set done!')

print('Preprocessing test set...')
test_df['clean_text'] = test_df['text'].apply(preprocess_text)
print('Test set done!')
```

```python
train_df[['text', 'clean_text', 'status']].head()
```

---
## 4. Feature Engineering

```python
tfidf = TfidfVectorizer(max_features=10000)

X_train_tfidf = tfidf.fit_transform(train_df['clean_text'])
X_test_tfidf  = tfidf.transform(test_df['clean_text'])

print('Train feature matrix shape:', X_train_tfidf.shape)
print('Test feature matrix shape :', X_test_tfidf.shape)
```

```python
le = LabelEncoder()
y_train = le.fit_transform(train_df['status'])
y_test  = le.transform(test_df['status'])

print('Label mapping:', dict(zip(le.classes_, le.transform(le.classes_))))
print('y_train shape:', y_train.shape)
print('y_test shape :', y_test.shape)
```

---
## 5. Classical ML Models

```python
results = {}

def evaluate_model(model_name, y_true, y_pred):
    print(f'=== {model_name} ===')
    print(f'Accuracy: {accuracy_score(y_true, y_pred):.4f}')
    print(classification_report(y_true, y_pred, target_names=le.classes_))
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=le.classes_, yticklabels=le.classes_)
    plt.title(f'Confusion Matrix — {model_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.tight_layout()
    plt.show()
    return round(accuracy_score(y_true, y_pred) * 100, 2)
```

```python
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train_tfidf, y_train)
lr_pred = lr_model.predict(X_test_tfidf)
results['Logistic Regression'] = evaluate_model('Logistic Regression', y_test, lr_pred)
```

```python
svc_model = LinearSVC(random_state=42, max_iter=2000)
svc_model.fit(X_train_tfidf, y_train)
svc_pred = svc_model.predict(X_test_tfidf)
results['SVC'] = evaluate_model('SVC', y_test, svc_pred)
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train_tfidf, y_train)
rf_pred = rf_model.predict(X_test_tfidf)
results['Random Forest'] = evaluate_model('Random Forest', y_test, rf_pred)
```

---
## 6. Deep Learning — ANN

```python
ann_model = Sequential([
    Input(shape=(X_train_tfidf.shape[1],)),
    Dense(512, activation='relu'),
    BatchNormalization(),
    Dropout(0.4),
    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(4, activation='softmax')
])

ann_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

ann_model.summary()
```

```python
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = dict(enumerate(class_weights))
print('Class weights:', class_weight_dict)

X_val_dense = X_test_tfidf.toarray()

BATCH_SIZE  = 256
EPOCHS      = 30
n_train     = X_train_tfidf.shape[0]
steps_epoch = math.ceil(n_train / BATCH_SIZE)

def train_generator():
    while True:
        idx    = np.random.permutation(n_train)
        X_shuf = X_train_tfidf[idx]
        y_shuf = y_train[idx]
        for start in range(0, n_train, BATCH_SIZE):
            end     = min(start + BATCH_SIZE, n_train)
            X_batch = X_shuf[start:end].toarray()
            y_batch = y_shuf[start:end]
            s_batch = np.array([class_weight_dict[label] for label in y_batch])
            yield X_batch, y_batch, s_batch

early_stop = EarlyStopping(monitor='val_loss', patience=5,
                           restore_best_weights=True, verbose=1)
reduce_lr  = ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                               patience=3, min_lr=1e-6, verbose=1)

history = ann_model.fit(
    train_generator(),
    steps_per_epoch = steps_epoch,
    epochs          = EPOCHS,
    validation_data = (X_val_dense, y_test),
    callbacks       = [early_stop, reduce_lr],
    verbose         = 1
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history.history['accuracy'],     label='Train Accuracy')
axes[0].plot(history.history['val_accuracy'], label='Val Accuracy')
axes[0].set_title('ANN — Accuracy')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].legend()

axes[1].plot(history.history['loss'],     label='Train Loss')
axes[1].plot(history.history['val_loss'], label='Val Loss')
axes[1].set_title('ANN — Loss')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
ann_pred_probs = ann_model.predict(X_val_dense)
ann_pred       = np.argmax(ann_pred_probs, axis=1)
results['ANN'] = evaluate_model('ANN', y_test, ann_pred)
```

## 8. Prediction Function (Random Forest)

```python
def preprocess_input(text):
    text   = text.lower()
    text   = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return ' '.join(tokens)

def predict_rf(text):
    cleaned = preprocess_input(text)
    vec     = tfidf.transform([cleaned])
    pred    = rf_model.predict(vec)[0]
    label   = le.inverse_transform([pred])[0]
    print(f'Predicted Label : {label}')
    return label
```

```python
samples = [
    "I feel so hopeless, I don't see a reason to continue living anymore.",
    "I have been feeling really sad and empty for weeks, nothing makes me happy.",
    "My heart is racing, I'm constantly worried and can't stop panicking.",
    "I had a great day today, went for a walk and met some friends!"
]

for sample in samples:
    print(f'Text: "{sample[:65]}..."')
    predict_rf(sample)
    print('-' * 65)
```

```python
user_text = input('Enter a text to classify: ')
print()
predict_rf(user_text)
```
